```python
from pathlib import Path
import pandas as pd

#  p-folder
cwd = Path.cwd()
project_root = cwd if (cwd / "data").exists() else cwd.parent

# dataset
file_path = project_root / "data" / "raw" / "superstore.csv"

# CSV file
df = pd.read_csv(file_path, encoding="utf-8-sig")

#  5 rows
df.head()
```

```python
# rows and columns
print("Dataset shape:", df.shape)

# column names
print("\nColumns:")
print(df.columns.tolist())

# missing values
print("\nMissing values:")
print(df.isnull().sum())

# duplicate r
print("\nDuplicate rows:", df.duplicated().sum())

# data-types
print("\nData types:")
print(df.dtypes)
```

```python
#duplicated rows
print("Duplicate rows:", df.duplicated().sum())

# Row IDs-duplicate
print("Duplicate Row IDs:", df["Row ID"].duplicated().sum())

# unique ord
print("Unique orders:", df["Order ID"].nunique())
```

```python
df.info()
```

```python
import pandas as pd

# original data
df_clean = df.copy()

# standardize
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace("-", "_", regex=False)
    .str.replace(" ", "_", regex=False)
)

# Convert dates by normalizing separators first
for col in ["order_date", "ship_date"]:

    dates = (
        df_clean[col]
        .astype(str)
        .str.strip()
        .str.replace("-", "/", regex=False)
    )

    # Handle single-digit and double-digit month/day
    df_clean[col] = pd.to_datetime(
        dates,
        format="%m/%d/%Y",
        errors="coerce"
    )

# Remove exact duplicates
df_clean = df_clean.drop_duplicates()

# Display results
df_clean[
    ["order_date", "ship_date", "sales", "profit"]
].head(10)
```

```python
print("Invalid order dates:", df_clean["order_date"].isna().sum())
print("Invalid ship dates:", df_clean["ship_date"].isna().sum())

print("\nFirst five dates:")
print(df_clean[["order_date", "ship_date"]].head())
```

```python
print("Total records:", len(df_clean))
print("Invalid order dates:", df_clean["order_date"].isna().sum())
print("Invalid ship dates:", df_clean["ship_date"].isna().sum())
print("Invalid quantities:", (df_clean["quantity"] <= 0).sum())
print("Negative sales:", (df_clean["sales"] < 0).sum())
print("Invalid discounts:", ((df_clean["discount"] < 0) | (df_clean["discount"] > 1)).sum())
print("Shipping before order:", (df_clean["ship_date"] < df_clean["order_date"]).sum())
```
